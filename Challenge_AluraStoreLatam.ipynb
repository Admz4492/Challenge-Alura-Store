### Importación de datos



```python
import pandas as pd

url = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_1%20.csv"
url2 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_2.csv"
url3 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_3.csv"
url4 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_4.csv"

tienda = pd.read_csv(url)
tienda2 = pd.read_csv(url2)
tienda3 = pd.read_csv(url3)
tienda4 = pd.read_csv(url4)

tienda.head()
```

#1. Análisis de facturación



```python
# Calcular facturación total por tienda
ingreso_tienda1 = tienda["Precio"].sum()
ingreso_tienda2 = tienda2["Precio"].sum()
ingreso_tienda3 = tienda3["Precio"].sum()
ingreso_tienda4 = tienda4["Precio"].sum()

# Mostrar resultados
print("Facturación Tienda 1: ", ingreso_tienda1)
print("Facturación Tienda 2: ", ingreso_tienda2)
print("Facturación Tienda 3: ", ingreso_tienda3)
print("Facturación Tienda 4: ", ingreso_tienda4)
```

```python
# Generacion de diccionario para facilitar futuras tareas
facturacion = {
    "Tienda 1": ingreso_tienda1,
    "Tienda 2": ingreso_tienda2,
    "Tienda 3": ingreso_tienda3,
    "Tienda 4": ingreso_tienda4
}

facturacion
```

# 2. Ventas por categoría

```python
# Tienda 1
categorias_t1 = tienda.groupby("Categoría del Producto")["Producto"].count().sort_values(ascending=False)

# Tienda 2
categorias_t2 = tienda2.groupby("Categoría del Producto")["Producto"].count().sort_values(ascending=False)

# Tienda 3
categorias_t3 = tienda3.groupby("Categoría del Producto")["Producto"].count().sort_values(ascending=False)

# Tienda 4
categorias_t4 = tienda4.groupby("Categoría del Producto")["Producto"].count().sort_values(ascending=False)

# Mostrar resultados
print("Tienda 1")
print(categorias_t1)

print("\nTienda 2")
print(categorias_t2)

print("\nTienda 3")
print(categorias_t3)

print("\nTienda 4")
print(categorias_t4)
```

# 3. Calificación promedio de la tienda


```python
# Promedio de calificación por tienda
valoracion_t1 = tienda["Calificación"].mean()
valoracion_t2 = tienda2["Calificación"].mean()
valoracion_t3 = tienda3["Calificación"].mean()
valoracion_t4 = tienda4["Calificación"].mean()

print("Valoración promedio Tienda 1:", round(valoracion_t1, 2))
print("Valoración promedio Tienda 2:", round(valoracion_t2, 2))
print("Valoración promedio Tienda 3:", round(valoracion_t3, 2))
print("Valoración promedio Tienda 4:", round(valoracion_t4, 2))
```

```python
valoraciones = {
    "Tienda 1": valoracion_t1,
    "Tienda 2": valoracion_t2,
    "Tienda 3": valoracion_t3,
    "Tienda 4": valoracion_t4
}

valoraciones
```

# 4. Productos más y menos vendidos

```python
# Tienda 1
ventas_t1 = tienda["Producto"].value_counts()

# Tienda 2
ventas_t2 = tienda2["Producto"].value_counts()

# Tienda 3
ventas_t3 = tienda3["Producto"].value_counts()

# Tienda 4
ventas_t4 = tienda4["Producto"].value_counts()
```

```python
print("Tienda 1")
print("Más vendido:", ventas_t1.idxmax(), "-", ventas_t1.max(), "ventas")
print("Menos vendido:", ventas_t1.idxmin(), "-", ventas_t1.min(), "ventas")

print("\nTienda 2")
print("Más vendido:", ventas_t2.idxmax(), "-", ventas_t2.max(), "ventas")
print("Menos vendido:", ventas_t2.idxmin(), "-", ventas_t2.min(), "ventas")

print("\nTienda 3")
print("Más vendido:", ventas_t3.idxmax(), "-", ventas_t3.max(), "ventas")
print("Menos vendido:", ventas_t3.idxmin(), "-", ventas_t3.min(), "ventas")

print("\nTienda 4")
print("Más vendido:", ventas_t4.idxmax(), "-", ventas_t4.max(), "ventas")
print("Menos vendido:", ventas_t4.idxmin(), "-", ventas_t4.min(), "ventas")
```

# 5. Envío promedio por tienda

```python
# Promedio de costo de envío por tienda
envio_t1 = tienda["Costo de envío"].mean()
envio_t2 = tienda2["Costo de envío"].mean()
envio_t3 = tienda3["Costo de envío"].mean()
envio_t4 = tienda4["Costo de envío"].mean()

print("Envío promedio Tienda 1:", round(envio_t1, 2))
print("Envío promedio Tienda 2:", round(envio_t2, 2))
print("Envío promedio Tienda 3:", round(envio_t3, 2))
print("Envío promedio Tienda 4:", round(envio_t4, 2))
```

6. Graficos importantes

```python
# Facturacion Total por Tienda
import matplotlib.pyplot as plt

facturacion = {
    "Tienda 1": tienda["Precio"].sum(),
    "Tienda 2": tienda2["Precio"].sum(),
    "Tienda 3": tienda3["Precio"].sum(),
    "Tienda 4": tienda4["Precio"].sum()
}

plt.figure()
plt.bar(facturacion.keys(), facturacion.values())

plt.title("Facturación Total por Tienda")
plt.xlabel("Tiendas")
plt.ylabel("Ingresos Totales")

plt.show()
```

```python
# Valoración Promedio por Tienda
valoraciones = {
    "Tienda 1": tienda["Calificación"].mean(),
    "Tienda 2": tienda2["Calificación"].mean(),
    "Tienda 3": tienda3["Calificación"].mean(),
    "Tienda 4": tienda4["Calificación"].mean()
}

plt.figure()

plt.plot(
    list(valoraciones.keys()),
    list(valoraciones.values()),
    marker='o'
)

plt.title("Valoración Promedio por Tienda")
plt.xlabel("Tiendas")
plt.ylabel("Calificación Promedio")
plt.ylim(0, 5)

plt.show()
```

```python
# Distribucion de Categorias (Tienda con menor facturacion - Tienda 4)

categorias_t4 = tienda4["Categoría del Producto"].value_counts()

plt.figure()
plt.pie(categorias_t4.values, labels=categorias_t4.index, autopct='%1.1f%%')

plt.title("Distribución de Categorías - Tienda 4")

plt.show()
```

**Informe Final – Análisis de Desempeño de Alura Store**

1. Introducción

  El presente informe tiene como objetivo analizar el desempeño de las cuatro tiendas de la cadena Alura Store, con el fin de recomendar cuál de ellas debería ser vendida para que el Sr. Juan pueda invertir en un nuevo emprendimiento.

  Para la toma de decisión se evaluaron los siguientes indicadores clave:

- Ingresos totales por tienda

- Categorías de productos más y menos vendidas

- Productos más y menos vendidos

- Calificación promedio de los clientes

- Costo de envío promedio

  A partir del análisis de datos y las visualizaciones generadas, se identificaron fortalezas y debilidades en cada tienda.

2. Desarrollo del análisis

  **2.1 Ingresos Totales**

  El análisis de facturación mostró diferencias claras entre las tiendas.
  La Tienda 4 presentó el menor ingreso total en comparación con las demás, lo que indica un menor volumen de ventas o menor ticket promedio.

  Dado que la facturación es uno de los indicadores más importantes de rentabilidad, este resultado posiciona a la Tienda 4 como una de las menos competitivas.

**2.2 Categorías y Productos Más Vendidos**

  Al analizar la distribución de categorías:

- La Tienda 4 mostró menor diversidad en sus ventas.

- Presenta alta dependencia de pocas categorías.

- Algunos productos tienen muy baja rotación.

  En comparación, las otras tiendas muestran una distribución más equilibrada y mayor volumen de ventas por categoría, lo que reduce el riesgo comercial.

  Además, los productos menos vendidos en la Tienda 4 reflejan baja demanda en ciertos segmentos, lo que puede indicar problemas de estrategia comercial o posicionamiento.

**2.3 Calificación Promedio de Clientes**

  La satisfacción del cliente es clave para la sostenibilidad del negocio.

  La Tienda 4 presentó una de las valoraciones promedio más bajas entre las cuatro tiendas. Aunque la diferencia puede no ser extrema, sí indica menor percepción de calidad o experiencia por parte de los clientes.

**2.4 Costo de Envío Promedio**

La Tienda 4 presenta el costo de envío promedio más bajo entre todas las tiendas. Esto representa una ventaja competitiva importante, ya que un menor costo de envío puede incentivar la decisión de compra y mejorar la percepción del cliente.

Sin embargo, a pesar de contar con el envío más económico, la Tienda 4 no logra compensar esta ventaja en otros indicadores clave como la facturación total y el desempeño general de ventas. Esto sugiere que el costo de envío no es el principal factor que está afectando su rendimiento.

3. Conclusión y Recomendación Final

Con base en los análisis realizados, se concluye que la Tienda 4 presenta el desempeño más bajo en términos generales.

En comparación con las demás tiendas, la Tienda 4 muestra debilidades financieras y comerciales.

Por lo tanto, se recomienda que el Sr. Juan venda la Tienda 4, ya que representa la unidad con menor eficiencia relativa dentro de la cadena. Esta decisión permitiría liberar recursos para invertir en un nuevo emprendimiento con mayor potencial de crecimiento.
